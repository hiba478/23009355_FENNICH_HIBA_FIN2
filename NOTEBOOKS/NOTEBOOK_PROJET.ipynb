# Problématique du projet

**Comment l'analyse des séries temporelles permet-elle d'étudier et de comparer l'évolution des taux de change des différentes monnaies asiatiques face au dollar américain, et d'identifier les monnaies les plus stables et les plus volatiles au fil du temps ?**

## Objectif de l'analyse

Ce projet vise à étudier l'évolution des taux de change des monnaies présentes dans le jeu de données par rapport au dollar américain (USD), à comparer leurs fluctuations et à mettre en évidence les différences de stabilité et de volatilité entre elles. L'analyse s'appuie sur les séries temporelles, les statistiques descriptives et les visualisations réalisées avec Python.

## Méthodologie adoptée

Pour réaliser ce travail pratique, nous avons adopté une démarche progressive basée sur l'étude et l'expérimentation de différents algorithmes de Machine Learning.

La méthodologie suivie repose sur les étapes suivantes :

1. **Présentation des algorithmes :** introduction de chaque algorithme, de son principe de fonctionnement et de ses principales caractéristiques.

2. **Illustration graphique :** présentation d'un schéma original permettant de mieux comprendre le fonctionnement de chaque méthode.

3. **Étude des cas d'utilisation :** identification de deux exemples concrets d'application pour chaque algorithme.

4. **Expérimentation avec Python :** mise en œuvre des algorithmes à l'aide des bibliothèques adaptées et observation des résultats obtenus.

5. **Analyse des résultats :** interprétation des sorties produites afin de comprendre le comportement de chaque algorithme.

6. **Étude des hyperparamètres :** modification d'un hyperparamètre pour observer son influence sur le fonctionnement et les résultats du modèle.

Cette démarche permet de découvrir les différentes approches du Machine Learning, de comprendre leurs applications pratiques et de mieux appréhender l'impact des paramètres sur les performances des modèles.

#### Since each currency has different missing values, each one should be checked separately and the missing values ​​of this dataset should not be corrected in general.

```python
def describe(df):
    """Return a Pandas dataframe describing the contents of a source dataframe including missing values."""

    variables = []
    dtypes = []
    count = []
    unique = []
    missing = []


    for item in df.columns:
        variables.append(item)
        dtypes.append(df[item].dtype)
        count.append(len(df[item]))
        unique.append(len(df[item].unique()))
        missing.append(df[item].isna().sum())


    output = pd.DataFrame({
        'variable': variables,
        'dtype': dtypes,
        'count': count,
        'unique': unique,
        'missing value': missing
    })

    return output
```
